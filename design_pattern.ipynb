## Proxy Design Pattern
### Authentication and Authorization layer to access a entity

```python
# Proxy
# When one class takes action on the behalf of other class or provides a gate to other class, that's called the proxy design pattern
```

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    
    @abstractmethod
    def drive(self):
        pass
    
    @abstractmethod
    def get_info(self):
        pass   
```

```python
class RealCar(Vehicle):
    def __int__(self, model_name, owner_name):
        self.model_name = model_name
        self.owner_name = owner_name
        self.is_engine_running= False

    def start_engine(self):
        if not self.is_engine_running:
            self.is_engine_running= True

    def drive(self):
        if self.is_engine_running:
            print('You can drive the car')
        else:
            print('You cant drive the car')

    def get_info(self):
        print(f'Model Name - {self.model_name}, Owner Name - {self.owner_name}, Engine Status - {self.is_engine_running}')
```
