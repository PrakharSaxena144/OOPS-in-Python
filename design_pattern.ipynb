## Proxy Design Pattern
### Authentication and Authorization layer to access a entity

```python
# Proxy
# When one class takes action on the behalf of other class or provides a gate to other class, that's called the proxy design pattern
```

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    
    @abstractmethod
    def drive(self):
        pass
    
    @abstractmethod
    def get_info(self):
        pass   
```

```python
class RealCar(Vehicle):
    def __init__(self, model_name, owner_name):
        self.model_name = model_name
        self.owner_name = owner_name
        self.is_engine_running = False

    def start_engine(self):
        if not self.is_engine_running:
            self.is_engine_running = True
            print('Engine Started....!')

    def drive(self):
        if self.is_engine_running:
            print('You can drive the car')
        else:
            print('You cant drive the car')

    def get_info(self):
        print(f'Model Name - {self.model_name}, Owner Name - {self.owner_name}, Engine Status - {self.is_engine_running}')
```

```python
# Whenever we need to create a new instance of real car you don't to do from RealCar,
# You only need to use this ProxyCar class to create the instance
class CarProxy(Vehicle):
    def __init__(self, model_name, owner_name, driver_name):
        self.model_name= model_name
        self.owner_name= owner_name
        self.driver_name= driver_name
        self._real_car= None    # initialize instance of a real car
        self.is_access_granted= False

    def _create_real_car(self):
        if self._real_car is None:
            self._real_car = RealCar(self.model_name, self.owner_name)
        return self._real_car

    def _grant_access(self):
        if self.owner_name == self.driver_name:
            self.is_access_granted = True
            return True
        return False

    def start_engine(self):
        car= self._create_real_car()
        car.start_engine()      # Or by self._create_real_car().start_engine()

    def drive(self):
        if self._grant_access():
            self._create_real_car().drive()
        else:
            print('Access denied: driver is not the owner.')

    def get_info(self):
        car = self._create_real_car()
        car.get_info()
```

```python
carProxy= CarProxy('Audi', 'Madhav', 'Prakhar')
carProxy.start_engine()
```
