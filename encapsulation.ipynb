```python
class Student:
    def __init__(self, name, age):
    # Encapsulation in Variables
        # Public variables- can be accessed outside the class
        self.name= name
        self.age= age

        # Private Variables
        self.__grades= []

        # Protected var (accessed from inherited classes)
        self._student_id= f'{self.name}&&{self.age}'

    # Encapsulation in funcs
    # Public Func
    def add_grades(self, grade):
        self.__grades.append(grade)
        print(f'Grade Added for {self.name}')

    def find_avg(self):
        avg= sum(self.__grades)/len(self.__grades)
        print(f'AVG - {avg}')
        return avg
    
    # Private Func
    def __is_passing(self) -> bool:
        return self.find_avg()>= 60
    
    # Protected Func
    def _find_gpa(self):
        gpa= self.find_avg()/4
        print(f'GPA - {gpa}')
        return gpa

    # Mix as well: Define logic behind Private Func, and the client facing func is a Public func
    def get_status(self):
        return self.__is_passing()
```

```python
student= Student('Aakash', 24)
```

```python
print(student.name)
print(student.age)
print(student._student_id)
print(student.__grades)     # Can't access Private variable
```

```python
student.add_grades(30)
student.find_avg()
student._find_gpa()
```

```python
student.__is_passing()      # Can't access Private Func
```

```python
student.get_status()        # Can access Private Func hidden behind a public func
```
