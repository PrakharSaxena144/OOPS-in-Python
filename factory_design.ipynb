```python
# Starting with a simple abstract class for vehicles and my other particular vechile use this abstract class
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, name):
        self.name= name
    
    @abstractmethod
    def start_engine():
        pass

    @abstractmethod
    def get_info():
        pass

class Car(Vehicle):
    def __init__(self, name, doors):
        self.name= name
        self.doors= doors

    def start_engine(self):
        print(f"Engine is started of {self.name}")
    
    def get_info(self):
        print(f"Info of {self.name}, no of doors: {self.doors}")

class Motorcycle(Vehicle):
    def __init__(self, name, cc):
        self.name= name
        self.cc= cc

    def start_engine(self):
        print(f"Engine is started of {self.name}")
    
    def get_info(self):
        print(f"Info of {self.name}, cc: {self.cc}")
```

```python
## Client Code (Creating objects for these classes)
## Factor Design logic helps us to encapsulate/decouple the object creation logic from the client
# here lets do user side input (can also do hardcode as in early files)
i= input("Enter the vehicle name: ")

if i=="car":
    object= Car(i, 4)
else:
    object1= Motorcycle(i, 100)
object.start_engine()
object.get_info()
```

```python
# class design to create new vehicle
class VehiclesFactory():
    @staticmethod
    def create_new_vehicle(name, **kwargs):
        if name=="car":
            doors= kwargs.get('doors', 4)
            return Car(name, doors)
        elif name=="car":
            cc= kwargs.get('cc', 200)
            return Motorcycle(name, cc)
```

```python
# Client Code
car= VehiclesFactory.create_new_vehicle('car', doors= 4)
car.get_info()
Motorcycle= VehiclesFactory.create_new_vehicle('motorcycle', cc=100)
Motorcycle.get_info()
```
