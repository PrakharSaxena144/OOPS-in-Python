```python
# Starting with a simple abstract class for vehicles and my other particular vechile use this abstract class
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, name):
        self.name= name
    
    @abstractmethod
    def start_engine():
        pass

    @abstractmethod
    def get_info():
        pass

class Car(Vehicle):
    def __init__(self, name, doors):
        self.name= name
        self.doors= doors

    def start_engine(self):
        print(f"Engine is started of {self.name}")
    
    def get_info(self):
        print(f"Info of {self.name}, no of doors: {self.doors}")

class Motorcycle(Vehicle):
    def __init__(self, name, cc):
        self.name= name
        self.cc= cc

    def start_engine(self):
        print(f"Engine is started of {self.name}")
    
    def get_info(self):
        print(f"Info of {self.name}, cc: {self.cc}")
```

```python
## Client Code (Creating objects for these classes)
## Factor Design logic helps us to encapsulate/decouple the object creation logic from the client
# here lets do user side input (can also do hardcode as in early files)
i= input("Enter the vehicle name: ")

if i=="car":
    object= Car(i, 4)
else:
    object= Motorcycle(i, 100)
object.start_engine()
object.get_info()
```
